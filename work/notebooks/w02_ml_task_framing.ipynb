## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could flag pages using a single threshold, such as low CTR or poor average position. However, page performance depends on multiple features together, including impressions, CTR, average position, and content age. An ML model can learn patterns across these features and rank pages for review instead of relying on manually chosen thresholds.

The starter pipeline provides an initial indication of this advantage: the hand-written rule achieved 0.24 Precision@50, while the random forest achieved 0.74 Precision@50 on the starter evaluation. This suggests that the learned model can produce a more useful refresh queue, although this result is specific to the provided evaluation setup.
